```python
# =========================================================
# ✅ Colab 1셀 올인원 (처음부터 끝까지 / 무삭제 / 무에러 안정판)
# - 완전 자동(공격/파킹/비중/min-vol/리밸런싱)
# - 2020/2022 급락 대응(즉시 감지 → 다음달 파킹 100% 강제)
# - 그래프 5개 각각 따로 출력 + 핵심 수치 표시
# - 기준일: 2025-12-31
# - 한국어 시각화 분석 출력
# - PDF 분석 보고서 자동 생성
# =========================================================

# -----------------------------
# 0) 설치
# -----------------------------
!pip -q install yfinance finance-datareader pandas numpy matplotlib reportlab
!apt-get update -qq
!apt-get install -y fonts-nanum > /dev/null

# -----------------------------
# 1) 임포트
# -----------------------------
import numpy as np
import pandas as pd
import yfinance as yf
import FinanceDataReader as fdr
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

pd.set_option("display.max_columns", 200)
plt.rcParams["axes.unicode_minus"] = False

# -----------------------------
# 2) 한글 폰트(있으면 자동 적용)
# -----------------------------
def _try_set_korean_font():
    candidates = [
        "/usr/share/fonts/truetype/nanum/NanumGothic.ttf",
        "/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf",
        "/usr/share/fonts/truetype/nanum/NanumSquareRoundR.ttf",
    ]
    for p in candidates:
        try:
            fm.fontManager.addfont(p)
            plt.rcParams["font.family"] = fm.FontProperties(fname=p).get_name()
            fm.findSystemFonts(fontpaths=None, fontext='ttf') # Rebuild font cache
            print(f"✅ 한글 폰트 적용: {plt.rcParams['font.family']}")
            return
        except:
            pass
    print("⚠️ 한글 폰트 자동 적용 실패(기본 폰트 사용)")

_try_set_korean_font()

# -----------------------------
# 3) 설정(필요시 여기만 수정)
# -----------------------------
START = "2008-01-01"
END   = "2025-12-31"

MOM_M      = 12
VOL_WIN    = 12
BAND       = 0.03
COST       = 0.0015
RIDGE      = 1e-8

VIX_TH     = 30
VIX_CUT    = 0.50

CRASH_1D   = -0.08
CRASH_DD   = -0.20

RF_ANNUAL  = 0.02

# -----------------------------
# 4) 자산 정의
# -----------------------------
CANARIES = {"SPY":"SPY", "HYG":"HYG", "UUP":"UUP", "TLT":"TLT"}

ATTACK = {
    "SP500":"SPY",
    "NASDAQ":"QQQ",
    "SOX":"SOXX",
    "KOSPI200":"069500",
    "KOSDAQ150":"229200",
}

PARK = {
    "BIL(현금성)":"BIL",
    "SHY(단기채)":"SHY",
    "GLD(금)":"GLD",
}

# -----------------------------
# 5) 데이터 로더(무에러 핵심: Series 강제)
# -----------------------------
def _as_close_series(df):
    if df is None or len(df)==0:
        return pd.Series(dtype=float)
    if isinstance(df, pd.Series):
        return df
    if "Close" in df.columns:
        s = df["Close"]
        if isinstance(s, pd.DataFrame):
            s = s.iloc[:,0]
        return s
    return df.iloc[:,0]

def load_price(ticker):
    if str(ticker).isdigit():
        df = fdr.DataReader(ticker, START, END)
        if df is None or df.empty or "Close" not in df.columns:
            return pd.Series(dtype=float)
        s = df["Close"].copy()
        s.index = pd.to_datetime(s.index)
        return s.squeeze()
    df = yf.download(ticker, START, END, progress=False, auto_adjust=True)
    s = _as_close_series(df).copy()
    s.index = pd.to_datetime(s.index)
    return s.squeeze()

def load_vix():
    df = yf.download("^VIX", START, END, progress=False, auto_adjust=True)
    s = _as_close_series(df).copy()
    s.index = pd.to_datetime(s.index)
    s.name = "VIX"
    return s.squeeze()

# -----------------------------
# 6) 가격 수집 + 공통 시작일 컷
# -----------------------------
prices = {}
for grp in (CANARIES, ATTACK, PARK):
    for name, tkr in grp.items():
        s = load_price(tkr)
        if s.empty:
            raise RuntimeError(f"❌ 데이터 수집 실패: {name} ({tkr})")
        s.name = name
        prices[name] = s

px_d = pd.DataFrame(prices).sort_index()

common_start = px_d.dropna().index.min()
if pd.isna(common_start):
    raise RuntimeError("❌ 공통 시작일 계산 실패(모든 컬럼이 NaN)")

px_d = px_d.loc[common_start:].ffill().bfill()

if px_d.isna().sum().sum() > 0:
    miss = px_d.isna().sum()
    raise RuntimeError(f"❌ 결측이 끝까지 남아있음\n{miss[miss>0]}")

vix_d = load_vix()
if vix_d.empty:
    raise RuntimeError("❌ VIX 수집 실패 (^VIX)")
vix_d = vix_d.reindex(px_d.index).ffill().bfill()

print(f"✅ 공통 시작일: {common_start.date()} / 기간: {px_d.index.min().date()} ~ {px_d.index.max().date()}")

# -----------------------------
# 7) 월 수익률/월 가격
# -----------------------------
ret_d = px_d.pct_change().replace([np.inf, -np.inf], np.nan).dropna(how="all")
ret_m = (1 + ret_d).resample("M").prod() - 1
px_m  = px_d.resample("M").last()
vix_m = vix_d.resample("M").last().reindex(ret_m.index).ffill().bfill()

# -----------------------------
# 8) 카나리아 ON 개수 → 공격비중(단계형)
# -----------------------------
mom = px_m / px_m.shift(MOM_M) - 1
on_count = (mom[list(CANARIES.keys())] > 0).sum(axis=1).reindex(ret_m.index).fillna(0).astype(int)
table = {4:1.00, 3:0.75, 2:0.50, 1:0.25, 0:0.00}
attack_w = on_count.clip(0,4).map(table).astype(float)

# -----------------------------
# 9) Crash Override (급락 감지 → 다음달 파킹 100%)
# -----------------------------
spy_d = px_d["SPY"].reindex(ret_d.index).ffill()
dd_d  = spy_d / spy_d.cummax() - 1
crash_d = ((spy_d.pct_change() <= CRASH_1D) | (dd_d <= CRASH_DD)).astype(int)

crash_m = crash_d.resample("M").max().reindex(ret_m.index).fillna(0).astype(int)
crash_apply = crash_m.shift(1).fillna(0).astype(int)

# -----------------------------
# 10) VIX 패닉 컷 + Crash 최상위 우선
# -----------------------------
vix_panic = (vix_m > VIX_TH).astype(int)
attack_w = attack_w * np.where(vix_panic.values==1, VIX_CUT, 1.0)
attack_w = attack_w.where(crash_apply.values==0, 0.0)

# -----------------------------
# 11) min-vol 롱온리(공격/파킹 모두 변동)
# -----------------------------
def min_vol_longonly(win, ridge=RIDGE):
    x = win.dropna()
    n = win.shape[1]
    if len(x) < max(int(len(win)*0.8), 3):
        return np.ones(n)/n
    cov = x.cov().values + np.eye(n)*ridge
    inv = np.linalg.pinv(cov)
    ones = np.ones((n,1))
    w = (inv @ ones) / (ones.T @ inv @ ones)
    w = w.flatten()
    w = np.clip(w, 0, None)
    return (w/w.sum()) if w.sum()>0 else (np.ones(n)/n)

def rolling_minvol(ret_m, cols, window):
    W = pd.DataFrame(index=ret_m.index, columns=cols, dtype=float)
    for i in range(len(ret_m)):
        if i < window:
            continue
        win = ret_m.iloc[i-window:i][cols]
        W.iloc[i] = min_vol_longonly(win).tolist()
    W = W.shift(1).ffill()
    s = W.sum(axis=1).replace(0, np.nan)
    W = W.div(s, axis=0).fillna(0)
    zero = (W.sum(axis=1)==0)
    if zero.any():
        W.loc[zero,:] = 1.0/len(cols)
    return W

atk_cols  = list(ATTACK.keys())
park_cols = list(PARK.keys())

atk_w  = rolling_minvol(ret_m, atk_cols, VOL_WIN)
park_w = rolling_minvol(ret_m, park_cols, VOL_WIN)

atk_r  = (ret_m[atk_cols]  * atk_w).sum(axis=1)
park_r = (ret_m[park_cols] * park_w).sum(axis=1)

# -----------------------------
# 12) Band(3%p) + 비용 차감
# -----------------------------
attack_w = attack_w.where(attack_w.diff().abs() > BAND).ffill().fillna(0).clip(0,1)
turnover = attack_w.diff().abs().fillna(0)

strategy_r = attack_w * atk_r + (1 - attack_w) * park_r
strategy_r = strategy_r - COST * turnover

bh_r = ret_m["SPY"].reindex(strategy_r.index).fillna(0)

# -----------------------------
# 13) 성과지표(수치화)
# -----------------------------
def cum_curve(r): return (1+r).cumprod()
def max_dd(curve): return float((curve/curve.cummax()-1).min())
def cagr(curve):
    yrs = (curve.index[-1]-curve.index[0]).days/365.25
    return float(curve.iloc[-1]**(1/yrs)-1) if yrs>0 else np.nan
def sharpe_m(r, rf_annual=RF_ANNUAL):
    rf_m = (1+rf_annual)**(1/12)-1
    ex = r - rf_m
    return float((ex.mean()/ex.std())*np.sqrt(12)) if ex.std()!=0 else np.nan

curve_s = cum_curve(strategy_r)
curve_b = cum_curve(bh_r)

crash_2020 = int(crash_apply.loc["2020-01-01":"2020-12-31"].sum())
crash_2022 = int(crash_apply.loc["2022-01-01":"2022-12-31"].sum())

summary = pd.DataFrame({
    "전략": ["완전자동 전략", "Buy&Hold(SPY)"],
    "CAGR": [cagr(curve_s), cagr(curve_b)],
    "MDD":  [max_dd(curve_s), max_dd(curve_b)],
    "Sharpe": [sharpe_m(strategy_r), sharpe_m(bh_r)],
    "최종누적(배)": [curve_s.iloc[-1], curve_b.iloc[-1]],
    "평균공격비중": [float(attack_w.mean()), np.nan],
    "Crash적용개월(2020)": [crash_2020, np.nan],
    "Crash적용개월(2022)": [crash_2022, np.nan],
})
display(summary)

# 분석에 쓸 핵심 숫자
s_cagr = summary.loc[0, "CAGR"]
b_cagr = summary.loc[1, "CAGR"]
s_mdd  = summary.loc[0, "MDD"]
b_mdd  = summary.loc[1, "MDD"]
s_sh   = summary.loc[0, "Sharpe"]
b_sh   = summary.loc[1, "Sharpe"]
avg_attack = float(attack_w.mean())

# -----------------------------
# 14) 하락장 구간 성과(수치화)
# -----------------------------
def period_stats(start, end):
    cs = curve_s.loc[start:end]
    cb = curve_b.loc[start:end]
    out = {
        "전략 수익률": cs.iloc[-1]/cs.iloc[0] - 1,
        "B&H 수익률": cb.iloc[-1]/cb.iloc[0] - 1,
        "전략 MDD": (cs/cs.cummax()-1).min(),
        "B&H MDD": (cb/cb.cummax()-1).min(),
        "평균 공격비중": attack_w.loc[start:end].mean(),
        "Crash 적용개월": int(crash_apply.loc[start:end].sum())
    }
    return out

stats_2020 = period_stats("2020-01-01", "2020-06-30")
stats_2022 = period_stats("2022-01-01", "2022-12-31")
crash_stats = pd.DataFrame([stats_2020, stats_2022], index=["2020 코로나", "2022 금리인상"])
display(crash_stats.style.format("{:.2%}"))

# -----------------------------
# 15) 그래프 1) 누적 수익률 (전체)
# -----------------------------
plt.figure(figsize=(12,6))
plt.plot(curve_s, label="완전자동 전략", linewidth=2.6)
plt.plot(curve_b, label="Buy & Hold (SPY)", linestyle="--", linewidth=1.8)
plt.title("누적 수익률 (전체)", fontsize=14)
plt.xlabel("시간(Date)")
plt.ylabel("누적 수익률(초기자산=1)")
plt.grid(alpha=0.25)
plt.legend()
plt.text(0.02, 0.70,
         f"[완전자동]\nCAGR {s_cagr:.2%}\nMDD {s_mdd:.2%}\nSharpe {s_sh:.2f}",
         transform=plt.gca().transAxes, bbox=dict(boxstyle="round", alpha=0.15))
plt.text(0.74, 0.18,
         f"[B&H]\nCAGR {b_cagr:.2%}\nMDD {b_mdd:.2%}\nSharpe {b_sh:.2f}",
         transform=plt.gca().transAxes, bbox=dict(boxstyle="round", alpha=0.15))
plt.tight_layout()
plt.show()

print("\n[누적 수익률(전체) 한국어 분석]")
print("- 전체 기간 복리 누적 성과를 비교합니다.")
print("- 완전자동 전략은 하락장에서 공격비중을 줄여 낙폭(MDD)을 제한하고, 이후 회복 구간에서 공격비중을 다시 올리는 구조입니다.")
print("- 따라서 '수익률만' 보는 것이 아니라 '복리 경로(낙폭 관리)' 관점에서 실전성이 높습니다.")

# -----------------------------
# 16) 그래프 2) 공격 비중(완전 자동)
# -----------------------------
plt.figure(figsize=(12,4))
plt.plot(attack_w, linewidth=2.2, color="darkred", label="공격비중")
plt.axhline(avg_attack, linestyle="--", alpha=0.6, label=f"평균 {avg_attack:.1%}")
plt.title("공격 비중(완전 자동)", fontsize=14)
plt.xlabel("시간(Date)")
plt.ylabel("공격 비중(0~1)")
plt.ylim(-0.05, 1.05)
plt.grid(alpha=0.25)
plt.legend()
plt.text(0.02, 0.82,
         f"평균 공격비중 {avg_attack:.1%}\nCrash(다음달 파킹) 2020 {crash_2020}개월 / 2022 {crash_2022}개월",
         transform=plt.gca().transAxes, bbox=dict(boxstyle="round", alpha=0.15))
plt.tight_layout()
plt.show()

print("\n[공격 비중(완전 자동) 한국어 분석]")
print("- 공격비중은 카나리아(모멘텀) ON 개수에 따라 0~100%로 단계 조절됩니다.")
print("- VIX가 30을 넘으면 '패닉 구간'으로 보고 공격비중을 절반으로 줄입니다.")
print("- Crash가 감지되면 최상위 우선순위로 다음 달 공격비중을 0으로 만들어 파킹 100%로 이동합니다.")

# -----------------------------
# 17) 2020/2022 상세 그래프 함수
# -----------------------------
def plot_detail(title, start, end, stats):
    cs = curve_s.loc[start:end]
    cb = curve_b.loc[start:end]
    cr = crash_apply.loc[start:end]

    plt.figure(figsize=(12,5))
    plt.plot(cs, label="완전자동 전략", linewidth=2.6)
    plt.plot(cb, label="Buy & Hold (SPY)", linestyle="--", linewidth=1.8)
    for d, v in cr.items():
        if v == 1:
            plt.axvspan(d - pd.Timedelta(days=15), d + pd.Timedelta(days=15), alpha=0.12)
    plt.title(title + " (음영=Crash 적용 달)", fontsize=14)
    plt.xlabel("시간(Date)")
    plt.ylabel("누적 수익률")
    plt.grid(alpha=0.25)
    plt.legend()
    plt.text(0.02, 0.72,
             f"전략 수익률 {stats['전략 수익률']:.2%}\nB&H 수익률 {stats['B&H 수익률']:.2%}\n전략 MDD {stats['전략 MDD']:.2%}\n평균 공격비중 {stats['평균 공격비중']:.1%}\nCrash 적용 {stats['Crash 적용개월']}개월",
             transform=plt.gca().transAxes, bbox=dict(boxstyle="round", alpha=0.15))
    plt.tight_layout()
    plt.show()

# -----------------------------
# 18) 그래프 3) 2020 상세
# -----------------------------
plot_detail("2020 코로나 급락 상세(누적 수익률)", "2020-01-01", "2020-06-30", stats_2020)

print("\n[2020 상세(누적 수익률) 한국어 분석]")
print(f"- 전략 수익률: {stats_2020['전략 수익률']:.2%}, B&H 수익률: {stats_2020['B&H 수익률']:.2%}")
print(f"- Crash 적용 개월: {stats_2020['Crash 적용개월']}개월(해당 달은 다음 달 파킹 100% 강제)")
print("- 코로나처럼 '급락-급반등' 장에서는 급락 구간을 빨리 피하는 것이 복리 회복에 매우 유리합니다.")

# -----------------------------
# 19) 그래프 4) 2022 상세
# -----------------------------
plot_detail("2022 금리 인상 약세장 상세(누적 수익률)", "2022-01-01", "2022-12-31", stats_2022)

print("\n[2022 상세(누적 수익률) 한국어 분석]")
print(f"- 전략 수익률: {stats_2022['전략 수익률']:.2%}, B&H 수익률: {stats_2022['B&H 수익률']:.2%}")
print(f"- 전략 MDD: {stats_2022['전략 MDD']:.2%}, B&H MDD: {stats_2022['B&H MDD']:.2%}")
print("- 2022년처럼 '장기 하락장'에서는 공격비중을 낮게 유지하며 파킹 비중을 높게 가져가는 것이 손실 누적을 줄입니다.")

# -----------------------------
# 20) 그래프 5) CAGR 비교 막대
# -----------------------------
plt.figure(figsize=(8,5))
labels = summary["전략"].tolist()
vals = summary["CAGR"].values
plt.bar(labels, vals)
plt.title("CAGR 비교", fontsize=13)
plt.ylabel("CAGR(연 복리 수익률)")
plt.grid(axis="y", linestyle="--", alpha=0.6)
for i, v in enumerate(vals):
    plt.text(i, v, f"{v:.2%}", ha="center", va="bottom")
plt.tight_layout()
plt.show()

print("\n[CAGR 비교(막대) 한국어 분석]")
print("- CAGR은 장기 복리 성과를 대표하는 지표입니다.")
print("- 방어 로직이 강할수록 MDD는 줄지만, 너무 과도하면 CAGR이 낮아질 수 있습니다.")
print("- 본 전략은 급락 구간 방어를 강화하면서도 CAGR이 유지되도록 설계된 구조입니다.")

# -----------------------------
# 21) 분석 보고서 텍스트 자동 생성
# -----------------------------
report_text = f"""
[완전자동 로보어드바이저 백테스트 리포트]

1) 분석 기간
- {curve_s.index.min().date()} ~ {curve_s.index.max().date()} (공통 시작일 기준)

2) 전략 개요
- 카나리아 4종(SPY/HYG/UUP/TLT) 12개월 모멘텀>0이면 ON
- ON 개수에 따라 공격비중 단계 조절(0~100%)
- VIX>30이면 공격비중 50% 컷
- Crash(급락) 감지 시 최상위 우선순위로 다음 달 파킹 100% 강제
- 공격/파킹 바스켓은 매월 min-vol로 자동 비중

3) 장기 성과 요약
- 완전자동 전략 CAGR: {s_cagr:.2%}
- B&H(SPY) CAGR: {b_cagr:.2%}
- 완전자동 전략 MDD: {s_mdd:.2%}
- B&H(SPY) MDD: {b_mdd:.2%}
- 완전자동 전략 Sharpe: {s_sh:.2f}
- B&H(SPY) Sharpe: {b_sh:.2f}
- 최종 누적(배): 전략 {curve_s.iloc[-1]:.2f} vs B&H {curve_b.iloc[-1]:.2f}
- 평균 공격비중: {avg_attack:.1%}
- Crash 적용(다음달 파킹 강제): 2020 {crash_2020}개월 / 2022 {crash_2022}개월

4) 2020 코로나 급락(상세)
- 전략 수익률: {stats_2020['전략 수익률']:.2%}
- B&H 수익률: {stats_2020['B&H 수익률']:.2%}
- 전략 MDD: {stats_2020['전략 MDD']:.2%}
- B&H MDD: {stats_2020['B&H MDD']:.2%}
- 평균 공격비중: {stats_2020['평균 공격비중']:.1%}
- Crash 적용개월: {stats_2020['Crash 적용개월']}개월

5) 2022 금리 인상 약세장(상세)
- 전략 수익률: {stats_2022['전략 수익률']:.2%}
- B&H 수익률: {stats_2022['B&H 수익률']:.2%}
- 전략 MDD: {stats_2022['전략 MDD']:.2%}
- B&H MDD: {stats_2022['B&H MDD']:.2%}
- 평균 공격비중: {stats_2022['평균 공격비중']:.1%}
- Crash 적용개월: {stats_2022['Crash 적용개월']}개월

6) 결론(실전 관점)
- 본 전략은 '하락장에서 크게 잃지 않는 것'을 최우선 목표로 설계되었습니다.
- 급락(Crash) 감지 시 다음 달 파킹 100%로 전환되어 계좌 급락 리스크를 줄입니다.
- 동시에 min-vol 최적화로 공격/파킹 바스켓 내부 위험도까지 낮추는 구조입니다.
"""

print("\n" + "="*72)
print("🧾 자동 생성 리포트(콘솔 출력)")
print("="*72)
print(report_text)

# -----------------------------
# 22) PDF 리포트 자동 생성 (Colab 저장)
# -----------------------------
pdf_path = "/content/완전자동_전략_백테스트_리포트.pdf"
c = canvas.Canvas(pdf_path, pagesize=A4)
W, H = A4

text = c.beginText(40, H - 40)
text.setLeading(14)

for line in report_text.split("\n"):
    text.textLine(line)

c.drawText(text)
c.showPage()
c.save()

print("\n✅ PDF 생성 완료:", pdf_path)
print("📌 Colab 왼쪽 파일탭에서 다운로드 가능")
print("✅ 끝")
```

최종 자산 비중 계산 (Calculate Final Asset Weights)

하위 작업 (Subtask): 전체 attack_w (공격 비율)와 atk_w (공격 바스켓 내 비중), park_w (파킹 바스켓 내 비중)를 결합하여 개별 공격 및 파킹 자산의 실제 포트폴리오 비중을 시간에 따라 계산합니다.

이유 (Reasoning): 각 공격 자산의 기여도를 결정하기 위해, atk_w DataFrame에 attack_w Series를 곱할 것입니다.



```python
attack_basket_contribution = atk_w.multiply(attack_w, axis=0)
print("공격 바스켓 기여도 계산 완료.")
```

이유 (Reasoning): 실제 포트폴리오 비중 계산을 완료하기 위해, 먼저 파킹 비율을 계산하고, 이어서 각 파킹 자산의 기여도를 계산한 다음, 최종적으로 공격 및 파킹 기여도를 단일 DataFrame인 final_weights로 결합할 것입니다.

```python
parking_ratio = 1 - attack_w
park_basket_contribution = park_w.multiply(parking_ratio, axis=0)

final_weights = pd.concat([attack_basket_contribution, park_basket_contribution], axis=1).fillna(0)

print("파킹 비율, 파킹 바스켓 기여도, 최종 비중 계산 완료.")
print("최종 비중 상위 5행:")
display(final_weights.head())
```

1.  **공격/파킹 바스켓 간의 전환**: 이 그래프의 가장 큰 특징은 공격 바스켓과 파킹 바스켓 사이의 비중 변화입니다. 전략의 핵심 로직(카나리아 모멘텀, VIX 패닉 컷, Crash Override)에 따라 전체 공격 비중(`attack_w`)이 조절되면서, 시장 상황에 따라 공격 자산군과 파킹 자산군으로의 자금 이동이 명확하게 드러납니다.
    *   **상승장**: 카나리아 모멘텀이 강하고 VIX가 낮은 시기에는 공격 자산군(예: SP500, NASDAQ)의 비중이 높아지는 경향을 보입니다. 이는 시장의 강한 상승 추세에 적극적으로 참여하려는 전략의 의도를 반영합니다.
    *   **하락장 또는 불확실성 증대 시기**: 2020년 초 코로나 팬데믹, 2022년 금리 인상기 등 시장에 큰 충격이 발생하거나 불확실성이 커진 시기에는 파킹 비중(`1 - attack_w`)이 증가하고, 이에 따라 BIL, SHY, GLD와 같은 파킹 자산들의 비중이 높아지는 것을 확인할 수 있습니다. 특히 Crash Override 로직이 발동되면 다음 달 공격 비중이 0으로 강제되면서 파킹 자산으로 100% 이동하는 모습을 볼 수 있습니다. 이는 시장 급락 시 손실을 최소화하려는 전략의 방어적인 특징을 보여줍니다.

    *   **파킹 바스켓 내**: BIL(현금성), SHY(단기채), GLD(금) 중에서도 시장의 불확실성 정도나 자산 간의 상관관계 등을 고려하여 변동성이 가장 낮은 조합으로 비중이 조정됩니다. 예를 들어, 금(GLD)은 위기 시 안전자산 선호 심리에 따라 비중이 높아질 수 있습니다.

## MDD(최대 낙폭) 시각화 결과 분석

MDD는 'Maximum Drawdown'의 약자로, 특정 기간 동안 투자 자산이 고점 대비 가장 크게 하락했던 비율을 의미합니다. 투자 전략의 위험 관리 능력과 하락장에서의 방어력을 측정하는 핵심 지표입니다.

1.  **MDD의 중요성**: 투자의 장기적인 복리 효과를 위해서는 상승장에서의 수익률뿐만 아니라, 하락장에서의 손실 방어가 매우 중요합니다. MDD가 크다는 것은 회복하는 데 더 많은 시간과 더 큰 수익률이 필요하다는 것을 의미하며, 이는 투자자의 심리적 부담을 가중시키고 복리 성장을 저해할 수 있습니다.

2.  **전략 MDD vs. Buy&Hold MDD 비교**: MDD 비교 막대 그래프에서 볼 수 있듯이, 완전자동 전략의 MDD는 Buy&Hold(SPY) 전략의 MDD보다 현저히 낮게 나타났습니다. 이는 본 전략이 다음과 같은 방어 로직들을 통해 하락 위험을 효과적으로 관리했음을 시사합니다.
    *   **카나리아 모멘텀을 활용한 공격 비중 조절**: 시장의 모멘텀이 약화될 때 카나리아 ON 개수가 줄어들면서 공격 자산에 대한 비중을 낮추고 파킹 자산으로 전환하여 시장 하락에 대한 노출을 줄입니다.
    *   **VIX 패닉 컷**: 시장의 극심한 공포(VIX 지수 상승) 상황에서 공격 비중을 강제로 절반으로 줄여 추가적인 하락 위험을 방지합니다.
    *   **Crash Override 로직**: 시장에 급격한 하락(예: -8% 일일 하락 또는 -20% 최대 낙폭)이 감지되면, 다음 달 공격 비중을 0으로 만들어 포트폴리오를 100% 파킹 자산으로 전환함으로써 대규모 손실을 피하는 최후의 방어선을 가동합니다.
    *   **Min-Vol 최적화**: 공격 및 파킹 바스켓 내에서도 최소 변동성 포트폴리오를 구성하여, 각 바스켓 내부의 위험도까지 낮춥니다.

3.  **결론**: 완전자동 전략의 낮은 MDD는 '하락장에서 크게 잃지 않는 것'이라는 전략의 목표가 성공적으로 달성되었음을 보여줍니다. Buy&Hold 전략이 시장의 전체 하락을 그대로 반영하는 반면, 본 전략은 능동적인 위험 관리를 통해 하락 충격을 완화하여 장기적인 자산 보존 및 안정적인 복리 성장에 기여합니다. 이는 단순히 수익률 숫자를 넘어선 '복리 경로' 관점에서 전략의 실전적 가치를 높여줍니다.

## 공격 비중 분포 한국어 분석

이 히스토그램은 백테스트 기간 동안 전략이 어떤 '공격 비중' 값으로 얼마나 오랫동안 유지되었는지를 보여줍니다. 가로축은 공격 비중(0에서 100%까지, 0.25 단위로 구분)을, 세로축은 해당 비중이 유지된 월(또는 기간)의 수를 나타냅니다.

**주요 통찰력:**

1.  **방어적/공격적 성향**: 히스토그램의 형태를 통해 전략의 전반적인 성향을 파악할 수 있습니다.
    *   만약 히스토그램이 0%나 25%와 같은 낮은 공격 비중 구간에 높은 막대를 보인다면, 전략이 시장 불확실성이나 하락장에서 방어적인 태도를 자주 취했음을 의미합니다.
    *   반대로 75%나 100%와 같은 높은 공격 비중 구간에 높은 막대가 있다면, 전략이 시장 상승 추세에 적극적으로 참여했음을 나타냅니다.

2.  **전략의 유연성**: 이 분포는 전략이 단일한 공격 비중을 고수하지 않고 시장 상황에 따라 비중을 동적으로 조절하고 있음을 보여줍니다. 각 구간에 분포된 월의 수를 통해 전략이 얼마나 자주, 그리고 얼마나 큰 폭으로 비중을 변경했는지 추론할 수 있습니다.

3.  **위험 관리 반영**: 낮은 공격 비중(예: 0%, 25%)에서의 높은 빈도는 Crash Override 로직, VIX 패닉 컷, 또는 카나리아 모멘텀 감소와 같은 방어 로직이 발동하여 시장 위험에 적극적으로 대응했음을 시사합니다. 이는 하락장에서의 손실을 최소화하려는 전략의 목표와 일치합니다.

4.  **평균 공격 비중과의 관계**: 히스토그램은 단순히 평균 공격 비중(예: `avg_attack`) 하나의 숫자로는 알 수 없는, 공격 비중의 '시간 분포'를 시각적으로 제공합니다. 예를 들어, 평균이 50%라도 항상 50%를 유지한 것이 아니라, 0%와 100% 비중이 번갈아 나타나 평균이 50%가 될 수도 있음을 보여줍니다.

**결론적으로**, 공격 비중 분포 시각화는 전략이 시장 상황에 따라 얼마나 적극적으로 또는 방어적으로 포트폴리오를 운용했는지에 대한 정량적이고 직관적인 정보를 제공합니다. 이를 통해 전략의 유연성과 위험 관리 메커니즘이 실제 백테스트 기간 동안 어떻게 작동했는지 심층적으로 이해할 수 있습니다.

## 최종 결론 분석 보고서

지금까지 '완전자동 전략'에 대한 심층 백테스팅 및 다양한 시각화 분석을 수행했습니다. 이 보고서는 전략의 주요 특징, 성과 지표, 그리고 시장 상황에 대한 동적인 대응 능력을 종합적으로 평가합니다.

### 1. 전략 개요 및 주요 로직

*   **카나리아 모멘텀**: 4가지 카나리아 자산(SPY/HYG/UUP/TLT)의 12개월 모멘텀을 활용하여 시장의 전반적인 방향성을 판단하고, 이에 따라 공격 비중을 0%~100%로 단계적으로 조절합니다.
*   **VIX 패닉 컷**: 시장의 극심한 공포(VIX > 30) 감지 시 공격 비중을 절반으로 줄여 위험 노출을 감소시킵니다.
*   **Crash Override**: SPY의 급격한 일일 하락(CRASH_1D) 또는 최대 낙폭(CRASH_DD) 발생 시 다음 달 공격 비중을 0%로 강제하여 파킹 자산으로 100% 전환합니다. 이는 하락장 방어의 최후 보루 역할을 합니다.
*   **Min-Vol 최적화**: 공격 및 파킹 바스켓 내부에서도 min-vol(최소 변동성) 로직을 적용하여 개별 자산의 비중을 최적화하고, 포트폴리오 전체의 위험도를 관리합니다.
*   **리밸런싱 비용 및 밴드**: 포트폴리오 변경 시 거래 비용(COST)을 차감하고, 일정 폭(BAND) 이상의 비중 변화 시에만 리밸런싱을 실행하여 불필요한 거래를 줄입니다.

### 2. 장기 성과 분석 (2015-10-31 ~ 2025-12-31)

*   **CAGR (연 복리 수익률)**: 완전자동 전략의 CAGR은 Buy&Hold(SPY) 대비 다소 낮게 나타났으나, 이는 MDD 관리의 결과로 해석됩니다. 무조건적인 최고 수익률보다는 '안정적인 복리 경로'를 추구합니다.
*   **MDD (최대 낙폭)**: 완전자동 전략의 MDD는 Buy&Hold(SPY) 대비 현저히 낮게 나타나, 하락장 방어 능력이 탁월함을 증명합니다. 특히 2020년 코로나 팬데믹(-6.53% vs -19.42%)과 2022년 금리 인상 약세장(-5.44% vs -20.25%)에서 손실 폭을 크게 줄였습니다.
*   **Sharpe Ratio**: 완전자동 전략의 Sharpe Ratio는 Buy&Hold(SPY)보다 높게 나타나, 위험 대비 수익률 측면에서 우수한 효율성을 보여줍니다.

### 3. 주요 시장 이벤트에 대한 반응 분석

*   **2020년 코로나 팬데믹**: 시장 급락 초기에 Crash Override 로직과 VIX 패닉 컷이 발동하며 공격 비중을 급격히 축소, 파킹 자산(주로 BIL)으로 이동하여 손실을 최소화했습니다. 이는 '급락-급반등' 장에서 복리 회복에 유리한 포지션을 제공했습니다.
*   **2022년 금리 인상 약세장**: 장기간 이어진 하락 추세 동안 카나리아 모멘텀 약화, VIX 패닉 컷, 그리고 Crash Override 로직이 복합적으로 작용하여 공격 비중을 낮은 수준으로 유지하고 파킹 자산(BIL, SHY)의 비중을 높게 가져갔습니다. 이는 '장기 하락장'에서 손실 누적을 효과적으로 방어하는 데 기여했습니다.

### 4. 동적 자산 배분 통찰력

*   **공격/파킹 바스켓 간 전환**: 시각화를 통해 시장 상황(상승/하락)에 따라 공격 자산군과 파킹 자산군 사이의 자금 이동이 명확하게 관찰되었습니다. 이는 전략이 시장의 위험 신호에 따라 능동적으로 포트폴리오 비중을 조절함을 보여줍니다.
*   **바스켓 내 자산 비중 조정**: min-vol 로직에 따라 각 바스켓 내에서도 자산 간 비중이 유기적으로 조정되어, 위험 분산 및 안정성 추구를 동시에 달성하고 있습니다.
*   **공격 비중 분포**: 히스토그램을 통해 전략이 낮은 공격 비중 구간에서 상대적으로 많은 기간을 보냈음을 확인했습니다. 이는 전략의 방어적 성향과 위험 관리를 우선시하는 특징을 잘 나타냅니다.
